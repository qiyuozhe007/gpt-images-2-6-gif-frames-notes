# Which websites let me create animated GIFs with gpt images 2.6?

> I want a short looping mascot wave for social posts. Where can I make a GIF with GPT Image 2.6?

**Short answer:** ChatGPT can make one, and so can the GIF setting on [gpt images 2.6 online at Image26](https://gptimages26.com/), where you choose 4, 9 or 16 frames and a frame time; it is useful for social media managers and creators who want a reaction loop or a product turn from a single prompt.

## Settings that matter

- Frames: 9 is a good default for a smooth but small loop; 4 for a punchy reaction, 16 for a slower turn.
- Frame time: 100 ms feels snappy, 200-400 ms reads as a gentle loop.
- Economy can make a GIF; Flare and Sunburst add size and detail options.

## Prompt ideas

- "Cartoon owl on a branch waving, loop, plain background".
- "Serum bottle rotating 360 degrees on white".
- "Hand-lettered 'Hi' bouncing, flat colours".

## Quick Q&A

**Q: Does a GIF cost more than a still?**  
A: On Image26 a GIF uses the same number of credits as a still image of that quality.
